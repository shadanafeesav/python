```python
n=int(input("Enter number:"))
for i in range(1,n + 1):
    for j in range(1,i + 1):
        print(i*j , end=" ")
    print()
```

```python
square = lambda a:a*a
rectangle = lambda l,b:l*b
triangle= lambda b,h:0.5*b*h

while True:
    print("\n1.Square")
    print("2.Rectangle")
    print("3.Triangle")
    print("4.Exit")
    
    ch=int(input("Enter your choice:"))

    if ch==1:
        a=int(input("Enter value:"))
        print("Area of square:",square(a))
    elif ch==2:
        l=int(input("Enter length:"))
        b=int(input("Enter breadth:"))
        print("Area of Rectangle:",rectangle(l,b))    
    elif ch==3:
        b=int(input("Enter base:"))
        h=int(input("Enter height:"))
        print("Area of Rectangle:",triangle(b,h)) 
    elif ch==4:
        break
    else:
        print("Invalid choice")
        
```

```python
n=int(input("Enter a number:"))
print("Factors are:")
for i in range(1,n+1):
    if n%i==0:
        print(i)
```

```python
words=input("Enter words:").split()
longest=max(words,key=len)
print("Longest word:",longest)
print("Length:",len(longest))
```

```python
list1=list(map(int,input("Enter numbers:").split()))
for i in range(len(list1)):
    if list1[i]>100:
        list1[i]='over'
print("Result:",list1)
```

```python
list1=list(map(int,input("Enter First List:").split()))
list2=list(map(int,input("Enter Second List:").split()))

if len(list1)==len(list2):
    print("Lists are same length")
else:
    print("lists are not same length")
        
if sum(list1)==sum(list2):
    print("List have same sum")
else:
    print("list have not same sum")       

if set(list1)&set(list2):
    print("Common values are:",set(list1)&set(list2))
else:
    print("No common values")      
```
